```python
query = """
WITH customer_sales AS (
    SELECT
        "Customer ID",
        "Customer Name",
        SUM("Sales") AS total_sales
    FROM sales
    GROUP BY
        "Customer ID",
        "Customer Name"
),

customer_profit AS (
    SELECT
        "Customer ID",
        SUM("Profit") AS total_profit
    FROM sales
    GROUP BY "Customer ID"
)

SELECT
    cs."Customer Name",
    cs.total_sales,
    cp.total_profit
FROM customer_sales cs
JOIN customer_profit cp
    ON cs."Customer ID" = cp."Customer ID"
ORDER BY cs.total_sales DESC
LIMIT 10;
"""

pd.read_sql(query, engine)
```

```python
query = """
WITH customer_sales AS (
    SELECT
        "Customer ID",
        "Customer Name",
        "Segment",
        SUM("Sales") AS total_sales
    FROM sales
    GROUP BY
        "Customer ID",
        "Customer Name",
        "Segment"
)

SELECT
    "Customer Name",
    "Segment",
    total_sales,
    RANK() OVER (
        PARTITION BY "Segment"
        ORDER BY total_sales DESC
    ) AS segment_rank
FROM customer_sales
ORDER BY "Segment", segment_rank;
"""

pd.read_sql(query, engine)
```

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS month,
        SUM("Sales") AS total_sales
    FROM sales
    GROUP BY month
)

SELECT
    month,
    total_sales,
    LAG(total_sales) OVER (
        ORDER BY month
    ) AS previous_month_sales
FROM monthly_sales
ORDER BY month;
"""

pd.read_sql(query, engine)
```

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS month,
        SUM("Sales") AS total_sales
    FROM sales
    GROUP BY month
),

comparison AS (
    SELECT
        month,
        total_sales,
        LAG(total_sales) OVER (
            ORDER BY month
        ) AS previous_month_sales
    FROM monthly_sales
)

SELECT
    month,
    total_sales,
    previous_month_sales,
    total_sales - previous_month_sales AS sales_difference
FROM comparison
ORDER BY month;
"""

pd.read_sql(query, engine)
```

```python
query = """
CREATE VIEW IF NOT EXISTS customer_sales_summary AS
SELECT
    "Customer ID",
    "Customer Name",
    "Segment",
    SUM("Sales") AS total_sales,
    SUM("Profit") AS total_profit
FROM sales
GROUP BY
    "Customer ID",
    "Customer Name",
    "Segment";
"""

with engine.begin() as connection:
    connection.exec_driver_sql(query)

print("View created")
```

8️⃣ What is the monthly sales trend?

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS month,
        SUM("Sales") AS total_sales
    FROM sales
    GROUP BY month
)

SELECT
    month,
    total_sales,
    LAG(total_sales) OVER (
        ORDER BY month
    ) AS previous_month_sales
FROM monthly_sales
ORDER BY month;
"""

result = pd.read_sql(query, engine)
result
```

```python
query = """
WITH monthly_sales AS (
    SELECT
        strftime('%Y-%m', "Order Date") AS month,
        SUM("Sales") AS total_sales
    FROM sales
    GROUP BY month
),

comparison AS (
    SELECT
        month,
        total_sales,
        LAG(total_sales) OVER (
            ORDER BY month
        ) AS previous_month_sales
    FROM monthly_sales
)

SELECT
    month,
    total_sales,
    previous_month_sales,
    total_sales - previous_month_sales AS sales_change
FROM comparison
ORDER BY month;
"""

result = pd.read_sql(query, engine)
result
```

9️⃣ What is the average order value?

```python
query = """
SELECT
    COUNT(DISTINCT "Order ID") AS total_orders,
    SUM("Sales") AS total_sales,
    SUM("Sales") / COUNT(DISTINCT "Order ID") AS average_order_value
FROM sales;
"""

result = pd.read_sql(query, engine)
result
```

🔟 Which customers have high sales but low profit?

```python
query = """
SELECT
    "Customer ID",
    "Customer Name",
    SUM("Sales") AS total_sales,
    SUM("Profit") AS total_profit
FROM sales
GROUP BY
    "Customer ID",
    "Customer Name"
HAVING
    SUM("Sales") > (
        SELECT AVG(customer_sales)
        FROM (
            SELECT SUM("Sales") AS customer_sales
            FROM sales
            GROUP BY "Customer ID"
        )
    )
    AND SUM("Profit") < 0
ORDER BY total_sales DESC;
"""

result = pd.read_sql(query, engine)
result
```
